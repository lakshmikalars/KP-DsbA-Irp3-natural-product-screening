```python
from rdkit import Chem
import pandas as pd

csv_path = '/content/coconut_data/coconut_csv-08-2026.csv'

usecols = ['identifier', 'canonical_smiles']
df = pd.read_csv(csv_path, usecols=usecols)
print(f"Total rows loaded: {len(df)}")

def is_valid_smiles(smi):
    if pd.isna(smi):
        return False
    mol = Chem.MolFromSmiles(smi)
    return mol is not None

df['is_valid'] = df['canonical_smiles'].apply(is_valid_smiles)
invalid_count = (~df['is_valid']).sum()
print(f"Invalid/unparseable SMILES: {invalid_count}")

dup_count = df['canonical_smiles'].duplicated().sum()
print(f"Exact duplicate SMILES: {dup_count}")
```

```python
from rdkit import Chem
from rdkit.Chem import AllChem
from rdkit.DataStructs import ConvertToNumpyArray
import pandas as pd
import numpy as np
import time

# Reload cleaned dataset
csv_path = '/content/drive/MyDrive/Subtractive_Genomics_KP_Results/DsbA_Irp3_Screening_Pipeline/coconut_cleaned.csv'
df_clean = pd.read_csv(csv_path)
print(f"Loaded cleaned dataset: {len(df_clean)} rows")

n_bits = 2048
fp_array = np.zeros((len(df_clean), n_bits), dtype=np.uint8)
failed_indices = []

start = time.time()
for i, smi in enumerate(df_clean['canonical_smiles']):
    mol = Chem.MolFromSmiles(smi)
    if mol is None:
        failed_indices.append(i)
        continue
    fp = AllChem.GetMorganFingerprintAsBitVect(mol, radius=2, nBits=n_bits)
    arr = np.zeros((n_bits,), dtype=np.uint8)
    ConvertToNumpyArray(fp, arr)
    fp_array[i] = arr

    if (i + 1) % 100000 == 0:
        elapsed = time.time() - start
        print(f"{i+1}/{len(df_clean)} done ({elapsed/60:.1f} min elapsed)")

elapsed = time.time() - start
print(f"Total time: {elapsed/60:.1f} min")
print(f"Failed to fingerprint: {len(failed_indices)}")

# Save fingerprint array + identifiers to Drive
np.save('/content/drive/MyDrive/Subtractive_Genomics_KP_Results/DsbA_Irp3_Screening_Pipeline/fingerprints_2048.npy', fp_array)
df_clean[['identifier']].to_csv('/content/drive/MyDrive/Subtractive_Genomics_KP_Results/DsbA_Irp3_Screening_Pipeline/fingerprint_identifiers.csv', index=False)
print("Saved fingerprints and identifiers to Drive")
```

```python
import numpy as np
from scipy.spatial.distance import pdist

# pdist computes pairwise distances efficiently without building the full NxN matrix in memory
# 'jaccard' distance on binary vectors is mathematically equivalent to Tanimoto distance
tanimoto_distances = pdist(fp_sample, metric='jaccard')

print(f"Number of pairwise distances computed: {len(tanimoto_distances)}")
print(f"Mean pairwise Tanimoto distance: {tanimoto_distances.mean():.4f}")
print(f"Median pairwise Tanimoto distance: {np.median(tanimoto_distances):.4f}")
print(f"Std dev: {tanimoto_distances.std():.4f}")
print(f"Min: {tanimoto_distances.min():.4f}, Max: {tanimoto_distances.max():.4f}")
```

```python
from rdkit.Chem.Scaffolds import MurckoScaffold

def get_murcko_scaffold(smi):
    mol = Chem.MolFromSmiles(smi)
    if mol is None:
        return None
    scaffold = MurckoScaffold.GetScaffoldForMol(mol)
    return Chem.MolToSmiles(scaffold)

df_sample_diversity['murcko_scaffold'] = df_sample_diversity['canonical_smiles'].apply(get_murcko_scaffold)

n_unique_scaffolds = df_sample_diversity['murcko_scaffold'].nunique()
n_total = len(df_sample_diversity)
print(f"Unique Bemis-Murcko scaffolds: {n_unique_scaffolds} out of {n_total} compounds")
print(f"Scaffold diversity ratio: {n_unique_scaffolds/n_total:.4f}")
```

```python
import pandas as pd
from rdkit import Chem
from rdkit.Chem import AllChem
import os

df_final = pd.read_csv('/content/drive/MyDrive/Subtractive_Genomics_KP_Results/DsbA_Irp3_Screening_Pipeline/coconut_pains_filtered.csv')
benchmark_sample = df_final.sample(n=20, random_state=42).reset_index(drop=True)

os.makedirs('/content/ligands', exist_ok=True)

success_count = 0
for i, row in benchmark_sample.iterrows():
    mol = Chem.MolFromSmiles(row['canonical_smiles'])
    if mol is None:
        print(f"Failed to parse: {row['identifier']}")
        continue
    mol = Chem.AddHs(mol)
    embed_result = AllChem.EmbedMolecule(mol, randomSeed=42)
    if embed_result != 0:
        print(f"Failed to embed 3D: {row['identifier']}")
        continue
    AllChem.MMFFOptimizeMolecule(mol)
    sdf_path = f'/content/ligands/{row["identifier"]}.sdf'
    writer = Chem.SDWriter(sdf_path)
    writer.write(mol)
    writer.close()
    success_count += 1

print(f"Generated {success_count} / {len(benchmark_sample)} ligand SDF files")
print(f"Files in /content/ligands: {len(os.listdir('/content/ligands'))}")
```

```python
import time
import numpy as np
from rdkit import DataStructs
from rdkit.ML.Cluster import Butina

# Load the PAINS-filtered fingerprints
base = "/content/drive/MyDrive/Subtractive_Genomics_KP_Results/DsbA_Irp3_Screening_Pipeline"

fps_array = np.load(
    f"{base}/pains_filtered_fingerprints_2048.npy"
)

print("Full fingerprint array:", fps_array.shape)

# --------------------------------------------------
# BENCHMARK SAMPLE
# --------------------------------------------------
BENCHMARK_SIZE = 20000

rng = np.random.default_rng(42)
sample_indices = rng.choice(
    len(fps_array),
    size=BENCHMARK_SIZE,
    replace=False
)

sample_array = fps_array[sample_indices]

# Convert to RDKit ExplicitBitVect objects
sample_fps = []

for row in sample_array:
    bv = DataStructs.ExplicitBitVect(2048)
    on_bits = np.flatnonzero(row)

    for bit in on_bits:
        bv.SetBit(int(bit))

    sample_fps.append(bv)

print("Benchmark sample:", len(sample_fps))

# --------------------------------------------------
# BUTINA CLUSTERING BENCHMARK
# --------------------------------------------------
# Distance = 1 - Tanimoto similarity
# 0.4 distance = 0.6 Tanimoto similarity

DISTANCE_CUTOFF = 0.4

print(f"\nStarting Butina clustering...")
print(f"Distance cutoff: {DISTANCE_CUTOFF}")
print(f"Equivalent Tanimoto similarity cutoff: {1 - DISTANCE_CUTOFF:.2f}")

t0 = time.time()

# Calculate upper-triangle pairwise distances
dists = []

for i in range(1, len(sample_fps)):
    sims = DataStructs.BulkTanimotoSimilarity(
        sample_fps[i],
        sample_fps[:i]
    )

    dists.extend(1.0 - np.asarray(sims))

print(f"Pairwise distances calculated: {len(dists):,}")

# Perform Butina clustering
clusters = Butina.ClusterData(
    dists,
    len(sample_fps),
    DISTANCE_CUTOFF,
    isDistData=True
)

elapsed = time.time() - t0

print("\n--- CLUSTERING BENCHMARK ---")
print(f"Sample size: {len(sample_fps):,}")
print(f"Number of clusters: {len(clusters):,}")
print(f"Runtime: {elapsed/60:.2f} minutes")

if len(clusters) > 0:
    cluster_sizes = np.array([len(c) for c in clusters])

    print(f"Mean cluster size: {cluster_sizes.mean():.2f}")
    print(f"Median cluster size: {np.median(cluster_sizes):.2f}")
    print(f"Largest cluster: {cluster_sizes.max():,}")
    print(f"Singleton clusters: {(cluster_sizes == 1).sum():,}")

print("\nBenchmark complete.")
```

```python
from rdkit import Chem
from rdkit.Chem import AllChem
import os
import pandas as pd

# --------------------------------------------------
# FOLDERS
# --------------------------------------------------
input_dir = "/content/ligands_850"
os.makedirs(input_dir, exist_ok=True)

fail_log = []
success = 0

# --------------------------------------------------
# LIGAND PREPARATION
# --------------------------------------------------
for i, row in df_maxmin_850.iterrows():

    ident = str(row["identifier"])
    smi = str(row["canonical_smiles"])

    sdf_path = os.path.join(input_dir, f"{ident}.sdf")
    pdbqt_path = os.path.join(input_dir, f"{ident}.pdbqt")

    try:
        # 1. Parse SMILES
        mol = Chem.MolFromSmiles(smi)

        if mol is None:
            fail_log.append({
                "identifier": ident,
                "reason": "SMILES parsing failed"
            })
            continue

        # 2. Add hydrogens
        mol = Chem.AddHs(mol)

        # 3. Generate 3D structure
        embed_status = AllChem.EmbedMolecule(
            mol,
            randomSeed=42
        )

        if embed_status != 0:
            fail_log.append({
                "identifier": ident,
                "reason": "3D embedding failed"
            })
            continue

        # 4. MMFF optimization
        try:
            mmff_status = AllChem.MMFFOptimizeMolecule(mol)

        except Exception as e:
            fail_log.append({
                "identifier": ident,
                "reason": f"MMFF optimization failed: {str(e)[:150]}"
            })
            continue

        # 5. Write SDF
        writer = Chem.SDWriter(sdf_path)
        writer.write(mol)
        writer.close()

        # 6. Convert SDF → PDBQT using Open Babel
        exit_code = os.system(
            f'obabel "{sdf_path}" -O "{pdbqt_path}" 2>/dev/null'
        )

        # 7. Check conversion
        if (
            exit_code == 0
            and os.path.exists(pdbqt_path)
            and os.path.getsize(pdbqt_path) > 0
        ):
            success += 1

        else:
            fail_log.append({
                "identifier": ident,
                "reason": "Open Babel PDBQT conversion failed"
            })

    except Exception as e:

        fail_log.append({
            "identifier": ident,
            "reason": f"Unexpected error: {str(e)[:150]}"
        })

    # Progress display
    processed = i + 1

    if processed % 50 == 0:
        print(
            f"Processed {processed}/{len(df_maxmin_850)} | "
            f"Successful: {success} | "
            f"Failed: {len(fail_log)}"
        )


# --------------------------------------------------
# SAVE FAILURE LOG
# --------------------------------------------------
fail_df = pd.DataFrame(fail_log)

failure_file = "/content/drive/MyDrive/Subtractive_Genomics_KP_Results/DsbA_Irp3_Screening_Pipeline/ligand_preparation_failures.csv"

fail_df.to_csv(
    failure_file,
    index=False
)

# --------------------------------------------------
# FINAL SUMMARY
# --------------------------------------------------
print("\n========================================")
print("LIGAND PREPARATION COMPLETE")
print("========================================")

print(f"Total compounds: {len(df_maxmin_850)}")
print(f"Successfully converted: {success}")
print(f"Failed: {len(fail_log)}")

print(f"\nFailure log saved to:")
print(failure_file)

if len(fail_log) > 0:
    print("\nFailure reasons:")
    print(fail_df["reason"].value_counts())
```

```python
import time
import os
from vina import Vina

receptor = "/content/Irp3_receptor.pdbqt"
ligand_dir = "/content/drive/MyDrive/Subtractive_Genomics_KP_Results/DsbA_Irp3_Screening_Pipeline/ligands_850_pdbqt"

center = [-1.895, -4.660, 4.389]
box_size = [23.9, 25.4, 27.8]

ligand_files = sorted([
    f for f in os.listdir(ligand_dir)
    if f.endswith(".pdbqt")
])

# Benchmark only 20 ligands
test_files = ligand_files[:20]

print(f"Testing {len(test_files)} ligands on Irp3")

results = []
start_total = time.time()

for f in test_files:

    ligand_id = f.replace(".pdbqt", "")
    t0 = time.time()

    try:
        v = Vina(sf_name='vina')
        v.set_receptor(receptor)
        v.set_ligand_from_file(os.path.join(ligand_dir, f))

        v.compute_vina_maps(
            center=center,
            box_size=box_size
        )

        v.dock(
            exhaustiveness=8,
            n_poses=5
        )

        elapsed = time.time() - t0
        best_score = v.energies(n_poses=1)[0][0]

        results.append({
            "identifier": ligand_id,
            "time_sec": elapsed,
            "best_affinity": best_score
        })

        print(
            f"{ligand_id}: "
            f"{elapsed:.1f}s, "
            f"affinity={best_score:.2f} kcal/mol"
        )

    except Exception as e:
        print(f"FAILED: {ligand_id} — {e}")

total_time = time.time() - start_total

print("\n--- Irp3 Benchmark Summary ---")
print(f"Compounds docked: {len(results)} / {len(test_files)}")
print(f"Total time: {total_time/60:.2f} min")

if results:
    avg = sum(r["time_sec"] for r in results) / len(results)
    print(f"Average per compound: {avg:.2f} sec")
    print(
        f"Estimated time for 834 compounds: "
        f"{(avg * 834) / 3600:.2f} hours"
    )
```

```python
import subprocess
import os
import csv
import time
from pathlib import Path

base = "/content/drive/MyDrive/Subtractive_Genomics_KP_Results/DsbA_Irp3_Screening_Pipeline"
ligand_dir = f"{base}/ligands_850_pdbqt"

# --------------------------------------------------
# CONFIG — set per target before running
# --------------------------------------------------
TARGET_NAME = "DsbA"
RECEPTOR_PDBQT = f"{base}/DsbA_receptor.pdbqt"
CENTER = (-3.324, 1.442, 9.328)
SIZE = (26.1, 25.4, 30.0)

RESULTS_CSV = f"{base}/docking_results_{TARGET_NAME}.csv"
FAILURES_LOG = f"{base}/docking_failures_{TARGET_NAME}.csv"

MAX_COMPOUNDS_THIS_SESSION = 420
TIME_BUDGET_HOURS = None

# --------------------------------------------------
# SETUP: figure out what's already done
# --------------------------------------------------
completed_ids = set()
if os.path.exists(RESULTS_CSV):
    with open(RESULTS_CSV, "r") as f:
        reader = csv.DictReader(f)
        for row in reader:
            completed_ids.add(row["identifier"])
    print(f"Found existing results: {len(completed_ids)} compounds already done.")
else:
    with open(RESULTS_CSV, "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow(["identifier", "best_affinity_kcal_mol", "runtime_seconds"])
    print("No existing results file — starting fresh.")

if not os.path.exists(FAILURES_LOG):
    with open(FAILURES_LOG, "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow(["identifier", "error"])

all_ligands = sorted(Path(ligand_dir).glob("*.pdbqt"))
remaining_ligands = [lig for lig in all_ligands if lig.stem not in completed_ids]

print(f"Total ligands found: {len(all_ligands)}")
print(f"Remaining to dock: {len(remaining_ligands)}")

# --------------------------------------------------
# DOCKING LOOP
# --------------------------------------------------
session_start = time.time()
docked_this_session = 0

for lig_path in remaining_ligands:
    identifier = lig_path.stem

    if MAX_COMPOUNDS_THIS_SESSION and docked_this_session >= MAX_COMPOUNDS_THIS_SESSION:
        print(f"\nReached session compound limit ({MAX_COMPOUNDS_THIS_SESSION}). Stopping cleanly.")
        break
    if TIME_BUDGET_HOURS and (time.time() - session_start) / 3600 >= TIME_BUDGET_HOURS:
        print(f"\nReached session time budget ({TIME_BUDGET_HOURS}h). Stopping cleanly.")
        break

    out_pdbqt = f"/content/out_{identifier}.pdbqt"

    t0 = time.time()
    try:
        result = subprocess.run(
            [
                "vina",
                "--receptor", RECEPTOR_PDBQT,
                "--ligand", str(lig_path),
                "--center_x", str(CENTER[0]), "--center_y", str(CENTER[1]), "--center_z", str(CENTER[2]),
                "--size_x", str(SIZE[0]), "--size_y", str(SIZE[1]), "--size_z", str(SIZE[2]),
                "--out", out_pdbqt,
            ],
            capture_output=True, text=True, timeout=300
        )
        runtime = time.time() - t0

        if result.returncode != 0:
            raise RuntimeError(f"Vina exited with error: {result.stderr.strip()[:200]}")

        # Parse best affinity from stdout — first "mode 1" row after the header
        best_affinity = None
        for line in result.stdout.splitlines():
            stripped = line.strip()
            if stripped.startswith("1") and len(stripped.split()) >= 2:
                parts = stripped.split()
                try:
                    best_affinity = float(parts[1])
                    break
                except ValueError:
                    continue

        if best_affinity is None:
            raise ValueError(f"Could not parse affinity from stdout: {result.stdout[-300:]}")

        with open(RESULTS_CSV, "a", newline="") as f:
            writer = csv.writer(f)
            writer.writerow([identifier, best_affinity, round(runtime, 1)])

        docked_this_session += 1
        if docked_this_session % 10 == 0:
            elapsed_min = (time.time() - session_start) / 60
            print(f"[{docked_this_session}/{len(remaining_ligands)}] {identifier}: {best_affinity} kcal/mol "
                  f"({runtime:.1f}s) | session elapsed: {elapsed_min:.1f} min")

    except Exception as e:
        with open(FAILURES_LOG, "a", newline="") as f:
            writer = csv.writer(f)
            writer.writerow([identifier, str(e)])
        print(f"FAILED: {identifier} — {e}")

    if os.path.exists(out_pdbqt):
        os.remove(out_pdbqt)

print(f"\n--- SESSION SUMMARY ---")
print(f"Docked this session: {docked_this_session}")
print(f"Total completed overall: {len(completed_ids) + docked_this_session}")
print(f"Remaining: {len(all_ligands) - len(completed_ids) - docked_this_session}")
```

```python
import pandas as pd
import numpy as np

base = "/content/drive/MyDrive/Subtractive_Genomics_KP_Results/DsbA_Irp3_Screening_Pipeline"

# Load docking results for both targets
df_dsba = pd.read_csv(f"{base}/docking_results_DsbA.csv")
df_irp3 = pd.read_csv(f"{base}/docking_results_Irp3.csv")
df_dsba = df_dsba.rename(columns={'best_affinity_kcal_mol': 'DsbA_affinity'})
df_irp3 = df_irp3.rename(columns={'best_affinity_kcal_mol': 'Irp3_affinity'})

# Load the 850 library (has fingerprints via identifier match) + working library metadata
df_lib = pd.read_csv(f"{base}/final_working_library_MaxMin850.csv")

# Merge everything on identifier
df_merged = df_lib.merge(df_dsba[['identifier', 'DsbA_affinity']], on='identifier', how='left')
df_merged = df_merged.merge(df_irp3[['identifier', 'Irp3_affinity']], on='identifier', how='left')

print(f"Total compounds: {len(df_merged)}")
print(f"With DsbA score: {df_merged['DsbA_affinity'].notna().sum()}")
print(f"With Irp3 score: {df_merged['Irp3_affinity'].notna().sum()}")
print(f"With both scores: {df_merged[['DsbA_affinity','Irp3_affinity']].notna().all(axis=1).sum()}")
```

```python
# Get fingerprints for these 850 compounds — reload full fingerprint array + identifiers, then subset
fp_full = np.load(f"{base}/pains_filtered_fingerprints_2048.npy")
fp_ids = pd.read_csv(f"{base}/pains_filtered_fingerprint_identifiers.csv")
fp_ids['row_idx'] = range(len(fp_ids))

df_merged = df_merged.merge(fp_ids[['identifier', 'row_idx']], on='identifier', how='left')
df_merged = df_merged.dropna(subset=['row_idx'])
fp_subset = fp_full[df_merged['row_idx'].astype(int).values]

print(f"Fingerprints matched: {fp_subset.shape}")
```

```python
import umap
import matplotlib.pyplot as plt

reducer = umap.UMAP(n_components=2, random_state=42, metric='jaccard')
embedding = reducer.fit_transform(fp_subset)

df_merged['UMAP1'] = embedding[:, 0]
df_merged['UMAP2'] = embedding[:, 1]

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

sc1 = axes[0].scatter(df_merged['UMAP1'], df_merged['UMAP2'],
                       c=df_merged['DsbA_affinity'], cmap='viridis_r', s=20)
axes[0].set_title('Chemical Space — colored by DsbA affinity')
plt.colorbar(sc1, ax=axes[0], label='kcal/mol')

sc2 = axes[1].scatter(df_merged['UMAP1'], df_merged['UMAP2'],
                       c=df_merged['Irp3_affinity'], cmap='viridis_r', s=20)
axes[1].set_title('Chemical Space — colored by Irp3 affinity')
plt.colorbar(sc2, ax=axes[1], label='kcal/mol')

plt.tight_layout()
plt.savefig(f"{base}/chemical_space_score_overlay.png", dpi=150)
plt.show()
```

```python
df_both = df_merged.dropna(subset=['DsbA_affinity', 'Irp3_affinity']).copy()
print(f"Compounds with both scores: {len(df_both)}")

df_both['consensus_score'] = df_both[['DsbA_affinity', 'Irp3_affinity']].mean(axis=1)
df_both_sorted = df_both.sort_values('consensus_score')

print("\nTop 15 dual-target compounds (strongest average affinity):")
print(df_both_sorted[['identifier', 'DsbA_affinity', 'Irp3_affinity', 'consensus_score', 'parent_family']].head(15))

corr = df_both['DsbA_affinity'].corr(df_both['Irp3_affinity'])
print(f"\nCorrelation between DsbA and Irp3 affinity: {corr:.3f}")
```

```python
corr_mw = df_both['molecular_weight'].corr(df_both['consensus_score'])
corr_rotbonds = df_both['rotatable_bond_count'].corr(df_both['consensus_score'])
print(f"Correlation: MW vs consensus score: {corr_mw:.3f}")
print(f"Correlation: rotatable bonds vs consensus score: {corr_rotbonds:.3f}")
```

```python
if 'heavy_atom_count' not in df_both.columns:
    heavy_atoms = pd.read_csv(
        '/content/coconut_data/coconut_csv-08-2026.csv',
        usecols=['identifier', 'heavy_atom_count']
    )
    df_both = df_both.merge(heavy_atoms, on='identifier', how='left')

print(f"Missing heavy_atom_count after merge: {df_both['heavy_atom_count'].isna().sum()}")

df_both['DsbA_LE'] = -df_both['DsbA_affinity'] / df_both['heavy_atom_count']
df_both['Irp3_LE'] = -df_both['Irp3_affinity'] / df_both['heavy_atom_count']
df_both['consensus_LE'] = (df_both['DsbA_LE'] + df_both['Irp3_LE']) / 2

df_both_le_sorted = df_both.sort_values('consensus_LE', ascending=False)
print("\nTop 15 by ligand efficiency (size-corrected):")
print(df_both_le_sorted[['identifier', 'DsbA_affinity', 'Irp3_affinity', 'heavy_atom_count', 'consensus_LE', 'parent_family']].head(15))
```

```python
# Set a reasonable absolute affinity floor first (e.g., both scores at least -7 kcal/mol)
MIN_AFFINITY = -7.0

df_qualified = df_both[
    (df_both['DsbA_affinity'] <= MIN_AFFINITY) &
    (df_both['Irp3_affinity'] <= MIN_AFFINITY)
].copy()

print(f"Compounds meeting minimum affinity threshold ({MIN_AFFINITY} kcal/mol) on both targets: {len(df_qualified)}")

df_qualified_sorted = df_qualified.sort_values('consensus_LE', ascending=False)
print("\nTop 15: strong binders (≤-7 kcal/mol both targets), ranked by ligand efficiency:")
print(df_qualified_sorted[['identifier', 'DsbA_affinity', 'Irp3_affinity', 'heavy_atom_count', 'consensus_LE', 'parent_family']].head(15))
```

```python
import requests
import time

df_shortlist = pd.read_csv(f"{base}/shortlist_19_smiles_for_admet.csv")

results = []
for _, row in df_shortlist.iterrows():
    smi = row['canonical_smiles']
    ident = row['identifier']
    url = f"https://pubchem.ncbi.nlm.nih.gov/rest/pug/compound/smiles/{smi}/cids/JSON"
    try:
        resp = requests.get(url, timeout=10)
        if resp.status_code == 200:
            data = resp.json()
            cid = data['IdentifierList']['CID'][0]
        else:
            cid = None
    except Exception as e:
        cid = None
    results.append({'identifier': ident, 'canonical_smiles': smi, 'PubChem_CID': cid})
    time.sleep(0.3)  # be polite to the API, avoid rate-limiting

df_cid = pd.DataFrame(results)
print(df_cid)

df_cid.to_csv(f"{base}/shortlist_19_with_CID.csv", index=False)
print("\nCompounds with a PubChem CID found:", df_cid['PubChem_CID'].notna().sum(), "/ 19")
```

```python
# Treat CID=0 as missing, same as NaN
df_cid['PubChem_CID'] = df_cid['PubChem_CID'].replace(0, None)

print(f"Confirmed matches: {df_cid['PubChem_CID'].notna().sum()} / 19")
print("\nNo CID found for:")
print(df_cid[df_cid['PubChem_CID'].isna()][['identifier', 'canonical_smiles']])

df_cid.to_csv(f"{base}/shortlist_19_with_CID.csv", index=False)
```
